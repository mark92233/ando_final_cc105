```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('Coffee Shop Sales.csv')
```

```python
# Group by product category
grouped = df.groupby('product_category')

# Calculate the total number of unique products across the entire dataset
total_products = df['product_detail'].nunique()

print("=" * 55)
print(f"TOTAL UNIQUE PRODUCTS IN STORE: {total_products}")
print("=" * 55)
print("\n")

for category, cat_df in grouped:
    print("=" * 55)
    print(f"PRODUCT CATEGORY: {category}")
    print("=" * 55)
    
    # Get all unique prices for each specific product within this category
    product_prices = cat_df.groupby('product_detail')['unit_price'].unique()
    
    for i, (product, prices) in enumerate(product_prices.items(), 1):
        # Format prices nicely as currency (e.g., $3.50, $4.38)
        price_str = ", ".join([f"${p:.2f}" for p in prices])
        print(f"{i:2d}. {product:<35} | Price(s): {price_str}")
    
    print("\n")
```
